### Fake News Detection

```python
# import libraries
import pandas as pd 
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
import re
import string
```

```python
# load datasets
df_fake = pd.read_csv("E:/Data Analysis file/Fake news detection using Machine Learning/datasets/Fake.csv")
df_true = pd.read_csv("E:/Data Analysis file/Fake news detection using Machine Learning/datasets/True.csv")
```

```python
# Add class labels
df_fake["class"] = 0
df_true["class"] = 1
```

```python
# Select last 10 fake news records
df_fake_manual_testing = df_fake.tail(10)

# Remove those 10 rows from training data
for i in range(23480,23470,-1):
    df_fake.drop([i],axis = 0,inplace = True)

# Select last 10 true news records
df_true_manual_testing = df_true.tail(10)

# Remove those 10 rows from training data
for i in range(21416,21406,-1):
    df_true.drop([i],axis = 0,inplace = True)
```

```python
# Combine manual testing datasets
df_manual_testing = pd.concat([df_fake_manual_testing, df_true_manual_testing],axis = 0)

# Save manual testing data
df_manual_testing.to_csv("E:/Data Analysis file/Fake news detection using Machine Learning/datasets/manual_testing.csv")
```

```python
# Merge Fake and True datasets
df_marge = pd.concat([df_fake, df_true], axis = 0)
df_marge.head(10)
```

```python
df_marge.info
```

```python
# Remove unnecessary columns
df = df_marge.drop(["title","subject","date"], axis = 1)
df.head(10)
```

```python
# Shuffle the dataset - Merge karne ke baad pehle saari Fake News aur phir saari True News ho sakti hain. Shuffling unhe randomly mix kar deta hai.Uske baad random samples select karta hai.
df = df.sample(frac = 1)
```

```python
df.head(10)
```

```python
# Check missing values or null values
df.isnull().sum()
```

```python
# Text Cleaning Function
def word_drop(text):
    text = text.lower() # Convert text to lowercase
    text = re.sub(r'\[.*?\]', '', text) # Remove text inside square brackets
    text = re.sub(r'https?://\S+|www\.\S+', '', text) # Remove URLs
    text = re.sub(r'<.*?>+', '', text) # Remove HTML tags
    text = re.sub(r'[%s]' % re.escape(string.punctuation), '', text) # Remove punctuation
    text = re.sub(r'\n', ' ', text) # Replace newline with space
    text = re.sub(r'\w*\d\w*', '', text) # Remove words containing numbers
    return text # Return cleaned text
```

```python
# Apply function to every news article
df["text"] = df ["text"].apply(word_drop)
```

```python
df.head(10)
```

```python
# Separate input and output
x = df ["text"]
y = df ["class"]
```

```python
# Split data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.25)
```

```python
# Import TF-IDF Vectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# Create TF-IDF object
vectrorization = TfidfVectorizer()

#Fit and transform training data
xv_train = vectrorization.fit_transform(x_train)

#Transform testing data
xv_test = vectrorization.transform(x_test)
```

## Use Logistic Regression

```python
# Import Logistic Regression
from sklearn.linear_model import LogisticRegression
```

```python
# Create the model
LR = LogisticRegression()

# Train the model
LR.fit(xv_train, y_train)
```

```python
# Check model accuracy
LR.score(xv_test, y_test)
```

```python
# Make predictions
pred_LR = LR.predict(xv_test)
```

```python
# Generate classification report
print(classification_report(y_test, pred_LR))
```

#### Use Decision Tree Classification

```python
DT.score(xv_test, y_test)
```

```python
pred_DT = DT.predict(xv_test)
```

```python
print(classification_report(y_test, pred_DT))
```

#### Use Manual Testing

```python
# Convert numeric prediction into readable text
def output_lable(n):
    if n == 0:
        return "Fake News"
    elif n == 1:
        return "True News"

# Create manual testing function
def manual_testing(news):
    testing_news = {"text":[news]}  # Convert news into dictionary
    new_def_test = pd.DataFrame(testing_news)  # Convert dictionary into DataFrame
    new_def_test["text"] = new_def_test["text"].apply(word_drop) # Clean the entered news
    new_x_test = new_def_test["text"]  # Select cleaned text
    new_xv_test = vectrorization.transform(new_x_test) # Convert text into TF-IDF numbers

# Get predictions from all models
    pred_LR = LR.predict(new_xv_test)
    pred_DT = DT.predict(new_xv_test)
    pred_GBC = GBC.predict(new_xv_test)
    pred_RFC = RFC.predict(new_xv_test)

# Display all predictions   
    return print("\n\nLR Prediction: {} \nDT Prediction: {} \nGBC Prediction: {} \nRFC Prediction: {}".format(output_lable(pred_LR[0]),
                                                                                                              output_lable(pred_DT[0]),
                                                                                                              output_lable(pred_GBC[0]),
                                                                                                              output_lable(pred_RFC[0])))
                                                                                                            
```

```python
# Take input from user
news = str (input())
manual_testing(news)
```
